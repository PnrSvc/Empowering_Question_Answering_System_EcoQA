# Standalone model benchmark



```python
import pandas as pd
from pathlib import Path

ROOT=Path("/content/outputs")
frames=[]
for path in ROOT.glob("*/epoch_metrics.csv"):
    df=pd.read_csv(path)
    df.insert(0,"model",path.parent.name)
    frames.append(df)

benchmark=pd.concat(frames,ignore_index=True)
benchmark.to_csv("/content/model_benchmark.csv",index=False)
display(benchmark)
```
