# Confidence threshold calibration

```python
import json, pandas as pd
from pathlib import Path

P=Path("/content/validation_predictions.jsonl")
assert P.exists()

rows=[json.loads(x) for x in P.read_text().splitlines() if x.strip()]

result=[]
for tau in [0.70,0.80,0.85,0.90,0.95]:
    direct=[r for r in rows if float(r["confidence"])>=tau]
    precision=sum(bool(r["is_correct"]) for r in direct)/max(len(direct),1)
    coverage=len(direct)/len(rows)
    result.append({
        "threshold":tau,
        "direct_precision":precision,
        "coverage":coverage,
        "fallback_rate":1-coverage,
    })

pd.DataFrame(result)
```
