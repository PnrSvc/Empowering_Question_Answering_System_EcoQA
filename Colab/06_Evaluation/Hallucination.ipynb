# Hallucination Evaluation

Primary metric: response-level hallucination rate. Supplementary metric: atomic-claim-level hallucination rate.

```python
import pandas as pd
from pathlib import Path

P=Path("/content/hallucination_annotations.csv")
assert P.exists()

df=pd.read_csv(P)
required={"response_id","configuration","total_atomic_claims","unsupported_atomic_claims"}
assert required.issubset(df.columns)

df["hallucinated_response"]=(df["unsupported_atomic_claims"]>0).astype(int)

summary=[]
for config,g in df.groupby("configuration"):
    response_rate=100*g["hallucinated_response"].mean()
    claim_rate=100*g["unsupported_atomic_claims"].sum()/max(g["total_atomic_claims"].sum(),1)
    summary.append({
        "configuration":config,
        "responses":len(g),
        "hallucinated_responses":int(g["hallucinated_response"].sum()),
        "response_level_hallucination_rate":response_rate,
        "atomic_claim_hallucination_rate":claim_rate,
    })

pd.DataFrame(summary)
```
